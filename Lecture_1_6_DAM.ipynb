```python
#Missing Value
print(df.isna().sum())
```

```python
#Decision making structures
#If Statements
rating = df["Customer_Rating"].mean()

if rating >= 4:
  print("Customer satisfaction is good")
```

```python
revenue = df["Revenue_Lakh"].mean()

print("Average Revenue:", revenue)

if revenue > 300:
  print("Average revenue is high")
```

Lecture 3 Continuation of Byju's Case

```python
# Decision making structure
# Nested If Statement in Python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  print("Revenue is high")

  if rating >= 4:
    print("Customer satisfaction is also good")
  else:
    print("Customer satisfaction needs improvement")
else:
  print("Revenue is low")
```

```python
# For Loop
for rating in df["Customer_Rating"]:

  if rating >= 4:
      print("Good Rating")
  else:
      print("Needs Improvement")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()
#Making a decision based on the rating
if rating >= 4.5:
  print("Excellent")
elif rating >=4.0:
  print("Good")
elif rating >=3.0:
  print("Average")
else:
  print("Needs Improvement")
```

```python
# FOR Loop with range() iloc= integer location

for i in range(10):
    print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#WHILE Loop in Python
#while loop and i,j is variable to track
i = 0

while i < 5:
    revenue = df["Revenue_Lakh"].iloc[i]

    if revenue >= 300:
        print("High Revenue:", revenue)
    else:
        print("Low Revenue:", revenue)

    i = i + 1
```

```python
# Statistics for one important column: Final_Amount (what the customer paid)
amount = df["Final_Amount"]

print("Mean (average)    : Rs", round(amount.mean(), 2))
print("Median (middle)   : Rs", amount.median())
print("Mode (most common): Rs", amount.mode()[0])
print("Minimum           : Rs", amount.min())
print("Maximum           : Rs", amount.max())
print("Standard deviation : Rs", round(amount.std(), 2))
print("Range (max - min) : Rs", amount.max() - amount.min())
```

```python
# Comparison of Mean and Median

# Calculate average order amount
mean_value = amount.mean()

# Calculate middle order amount
median_value = amount.median()

print(f"Mean = Rs{mean_value:.2f}")
print(f"Median = Rs {median_value:.2f}")

# Compare mean and median
if mean_value > median_value:
  print("A few large orders may be pulling the average up.")
else:
  print("The average is close to or below the median")
```

Lecture 6: Data Visualaization

```python
# Quantile Plot for Normal Disribution
# Step 1: Import the libraries
import matplotlib.pyplot as plt
import scipy.stats as stats

# Step 2: Remove missing values from Final_Amount
order_amount = df["Final_Amount"].dropna()

# Step 3: Create the Q-Q plot
stats.probplot(order_amount, dist="norm", plot=plt)

# Step 4: Add a title
plt.title("Q-Q Plot if Final Order Amount")

# Step 5: Label the axes
plt.xlabel("Theoritical Quantiles")
plt.ylabel("Sample Quantiles")

# Step 6: Show the plot
plt.show()
```

```python
# BAR CHART
# Step 1: Find total revenue for each city
city_revenue = df.groupby("City")["Final_Amount"].sum()

# Step 2: Arrange cities from highest to lowest revenue
city_revenue = city_revenue.sort_values(ascending=False)

# Step 3: Print the revenue of each city
print(city_revenue)

# Step 4: Create a bar chart
city_revenue.plot(kind="bar")

# Step 5: Add chart title and y-axis label
plt.title("Revenue by City")
plt.ylabel("Total Revenue")

# Step 6: Show the plot
plt.show()
```

```python
# BAR CHART
# Step 1: Find total revenue for each city
city_revenue = df.groupby("Cuisine")["Final_Amount"].sum()

# Step 2: Arrange cities from highest to lowest revenue
city_revenue = city_revenue.sort_values(ascending=False)

# Step 3: Print the revenue of each city
print(city_revenue)

# Step 4: Create a bar chart
city_revenue.plot(kind="bar")

# Step 5: Add chart title and y-axis label
plt.title("Revenue by Cuisine")
plt.ylabel("Total Revenue")

# Step 6: Show the plot
plt.show()
```

```python
# LINE CHART
# Step 1: Convert Order_Date into a date format
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Step 2: Count the number of orders each month
monthly_orders = df.groupby(df["Order_Date"].dt.to_period('M'))["Order_ID"].count()

# Step 3: Print the monthly orders
print(monthly_orders)

# Step 4: Create a line chart
monthly_orders.plot(kind="line", marker="o")

# Step 5: Add chart title and y-axis label
plt.title("Monthly Demand Trend")
plt.ylabel("Number of Orders")
plt.xlabel("Month")

# Step 6: Show the plot
plt.show()
```

```python
# PIE CHART
# Step 1: Count each payment mode
payment_mode = df["Payment_Mode"].value_counts()

# Step 2: Create a pie chart
plt.pie(payment_mode, labels=payment_mode.index, autopct="%1.0f%%")

# Step 3: Add chart title
plt.title("Payment Mode Share")

# Step 4: Show the chart
plt.show()
```

```python
# PIE CHART
# Step 1: Count each Cuisine
Cuisine = df["Cuisine"].value_counts()

# Step 2: Create a pie chart
plt.pie(Cuisine , labels=Cuisine.index, autopct="%1.0f%%")

# Step 3: Add chart title
plt.title("Cuisine")

# Step 4: Show the chart
plt.show()
```

```python
# Histogram
# Create histogram of order amounts
plt.hist(df["Final_Amount"].dropna(), bins=10)

# Add title and labels
plt.title("Distribution of Order Amounts")
plt.xlabel("Final Order Amount (Rs)")
plt.ylabel("Number of Orders")

# Show the chart
plt.show()
```

```python
# Scatter Plot
# Create scatter plot
plt.scatter(df["Distance_Km"], df["Delivery_Time_Min"])

# Add title and labels
plt.title("Delivery Distance vs. Delivery Time")
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (Minutes)")

# Show chart
plt.show()
```

```python
# Heatmap: City x Cuisine porder counts (the dark/zero cells show gaps)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define the missing city_cuisine variable using a cross-tabulation
city_cuisine = pd.crosstab(df["City"], df["Cuisine"])

sns.heatmap(city_cuisine, annot=True, fmt="d", cmap="Reds")
plt.title("Orders: City x Cuisine (Seaborn)")
plt.show()
```
